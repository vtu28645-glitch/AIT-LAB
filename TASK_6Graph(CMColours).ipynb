```
def is_safe(vertex,color,graph, colours):
  for i in range (len(graph)):
    if graph [vertex][i]==1 and colours[i]==color:
      return False
  return True
def graph_coloring(vertex,m,graph,colours):
  if vertex==len(graph):
    return True
  for c0lor in range (1,m+1):
    if is_safe(vertex,c0lor,graph, colours):
      colours[vertex]=c0lor
      if graph_coloring(vertex+1,m,graph,colours):
        return True
      colours[vertex]=0
  return False
graph=[
    [0,1,1,0,0],#A
    [1,0,0,1,0],#B
    [1,0,0,1,0],#C
    [0,1,1,0,1],#D
    [0,0,0,1,0]#E
]
n=len(graph)
for m in range (1,n+1):
  colors = [0]*n
  if graph_coloring(0,m,graph,colors):
    print("Minimum number of colours:", m)
    print("colours assignment:")
    for i in range (n):
      print (chr(65+i),"->color",colors[i])
    break
```
