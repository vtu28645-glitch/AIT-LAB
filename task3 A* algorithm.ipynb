```
import heapq

def h(a, b):
  return abs(a[0] - b[0]) + abs(a[1] - b[1])

def astar(grid, start, goal):
  q = [(0, start)]
  g_cost = {start: 0}
  path_trace = {start: None}

  while q:
    f_cost, current = heapq.heappop(q)

    if current == goal:

      path = []
      while current is not None:
        path.append(current)
        current = path_trace[current]
      return path[::-1], g_cost[goal]


    for dx, dy in [(1, 0), (-1, 0), (0, 1), (0, -1)]:
      neighbor = (current[0] + dx, current[1] + dy)


      if (0 <= neighbor[0] < len(grid) and
          0 <= neighbor[1] < len(grid[0]) and
          grid[neighbor[0]][neighbor[1]] == 0):

        tentative_g_cost = g_cost[current] + 1

        if neighbor not in g_cost or tentative_g_cost < g_cost[neighbor]:
          g_cost[neighbor] = tentative_g_cost
          f_cost = tentative_g_cost + h(neighbor, goal)
          heapq.heappush(q, (f_cost, neighbor))
          path_trace[neighbor] = current

  return None, None


grid = [
    [0, 0, 0],
    [0, 1, 0],
    [0, 0, 0]
]
start = (0, 0)
end = (2, 2)

print("Input Grid:")
for row in grid:
  print(row)
print(f"Start: {start}")
print(f"Goal: {end}")

optimal_path, cost = astar(grid, start, end)

if optimal_path:
  print(f"Optimal Path: {optimal_path}")
  print(f"Cost: {cost}")
else:
  print("No optimal path found.")
```
